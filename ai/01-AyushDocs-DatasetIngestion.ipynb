using kaggle dataset    https://www.kaggle.com/datasets/vagifa/ethereum-frauddetection-dataset
which has a github mirror of https://raw.githubusercontent.com/Saket03-P/Ethereum-Transactions-Fraud-Detection/main/Datasets/transaction_dataset.csv

```python
import os
import re
import urllib.request
import numpy as np
import pandas as pd
```

```python
# Paths & sources (run this notebook from the `ai/` directory)
HERE = os.getcwd()
DATA_DIR = os.path.join(HERE, "data")
DATA_PATH = os.path.join(DATA_DIR, "transaction_dataset.csv")
MIRROR_URL = 'https://raw.githubusercontent.com/Saket03-P/Ethereum-Transactions-Fraud-Detection/main/Datasets/transaction_dataset.csv'
OUT_EXCEL = os.path.join(DATA_DIR, "transaction_ds.xlsx")


if os.path.exists(DATA_PATH):
    print(f"File already exists at {DATA_PATH}. Skipping download.")
else:
    os.makedirs(DATA_DIR, exist_ok=True)
    urllib.request.urlretrieve(MIRROR_URL, DATA_PATH)
    print(f"Downloaded file to {DATA_PATH}.")
```

```python
df = pd.read_csv(DATA_PATH,low_memory=False)
with pd.option_context("display.max_columns", None):
    display(df.head())
```

```python
print(df.shape)
print(list(df.columns))
```

```python
# Light typing so downstream notebooks never re-derive it.
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df["FLAG"] = pd.to_numeric(df["FLAG"], errors="coerce")
df = df[df["FLAG"].notna()].reset_index(drop=True)
df["FLAG"] = df["FLAG"].astype(int)

fraud = int((df["FLAG"] == 1).sum())
print(f"Labeled rows: {len(df):,}")
print(f"  Fraud (FLAG=1): {fraud:,}  ({fraud / len(df) * 100:.1f}%)")
print(f"  Legit (FLAG=0): {len(df) - fraud:,}  ({(len(df) - fraud) / len(df) * 100:.1f}%)")
df["FLAG"].value_counts().rename(index={1: "fraud", 0: "legit"})
```
